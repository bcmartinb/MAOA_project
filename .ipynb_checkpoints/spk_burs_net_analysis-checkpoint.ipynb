```python
def run_full_mea_analysis(
    df,
    assign_groups=True,
    use_group3=False,
    double_label=False,
    threshold_spikes_per_min=5
):
    """
    Master function that:
      1. Assigns Mutant/WT/GM groups (optional)
      2. Produces plate maps for all plates
      3. Plots ALL features (firing, burst, network, synchrony)
      4. Runs ALL stats and returns them

    Parameters
    ----------
    df : DataFrame
        MEA results dataframe
    assign_groups : bool
        Whether to create df['group3'] using assign_group3()
    use_group3 : bool
        Whether plate maps use group colors instead of cell_line
    double_label : bool
        Whether to draw both labels (group3 + cell_line)
    threshold_spikes_per_min : int
        Minimum threshold for firing rate electrodes
    """

    df = df.copy()

    # ----------------------------------------------------
    # 1. ASSIGN GROUPS (optional)
    # ----------------------------------------------------
    if assign_groups:
        df["group3"] = df["cell_line"].apply(assign_group3)


    # ----------------------------------------------------
    # 2. PRODUCE ALL PLATE MAPS
    # ----------------------------------------------------
    plates = sorted(df["plate_name"].unique())
    print("\n### GENERATING PLATE MAPS ###\n")
    for p in plates:
        print(f" → Plate map for {p}")
        plot_plate_map(
            df,
            plate_name=p,
            use_group3=use_group3,
            double_label=double_label
        )


    # ----------------------------------------------------
    # 3. GENERATE ALL PLOTS FOR ALL FEATURES
    # ----------------------------------------------------
    print("\n### GENERATING FEATURE PLOTS ###\n")

    features = ["firing", "burst", "network", "synchrony"]
    for feat in features:
        print(f"\n--- Plotting {feat.upper()} ---")
        plot_feature_boxplots(
            df,
            feature=feat,
            threshold_spikes_per_min=threshold_spikes_per_min
        )


    # ----------------------------------------------------
    # 4. RUN ALL STATS FOR ALL FEATURES
    # ----------------------------------------------------
    print("\n### RUNNING STATS ###\n")

    stats_out = {}
    for feat in features:
        print(f"\n--- Stats for {feat.upper()} ---")
        stats_out[feat] = run_feature_stats(
            df,
            feature=feat,
            threshold_spikes_per_min=threshold_spikes_per_min
        )

    print("\nAnalysis complete.\n")

    return {
        "stats": stats_out,
        "data_used": df
    }
```

```python
results = run_full_mea_analysis(
    df3_clean,
    assign_groups=True,
    use_group3=True,
    double_label=True
)
```
